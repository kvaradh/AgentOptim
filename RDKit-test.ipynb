# Week 0 -- Environment Sanity Check

Not part of the ADMET pipeline itself -- just confirms the `admet` conda environment (RDKit installed, kernel wired into the notebook) works before building anything on top of it: parse a SMILES string, render the molecule.

```python
from rdkit import Chem
from rdkit.Chem import Draw

# SMILES = a text string encoding molecular structure
aspirin_smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"

# Parse the SMILES into an RDKit molecule object
mol = Chem.MolFromSmiles(aspirin_smiles)

# Render it as an image
Draw.MolToImage(mol)
```
